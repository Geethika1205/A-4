```python
import SimpleITK as sitk
import numpy as np
import matplotlib.pyplot as plt

# Read CT
ct_img = sitk.ReadImage("/content/050.mha")
ct = sitk.GetArrayFromImage(ct_img)

# Read label
label_img = sitk.ReadImage("/content/050 (1).mha")
label = sitk.GetArrayFromImage(label_img)

# Find slices that actually contain labels
label_slices = np.where(np.any(label > 0, axis=(1, 2)))[0]

print("Number of labeled slices:", len(label_slices))
print("First labeled slice:", label_slices[0])
print("Last labeled slice:", label_slices[-1])

# Select middle labeled slice
slice_index = label_slices[len(label_slices) // 2]

print("Displaying slice:", slice_index)

# Display CT + label
plt.figure(figsize=(10, 8))

plt.imshow(ct[slice_index], cmap="gray")

# Display label only where label exists
masked_label = np.ma.masked_where(label[slice_index] == 0,
                                  label[slice_index])

plt.imshow(masked_label, cmap="jet", alpha=0.45)

plt.title("CT + Ground Truth Label")
plt.axis("off")
plt.colorbar(label="Label")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Convert CT to float
ct_float = ct.astype(np.float32)

# Bone window
bone_min = 150
bone_max = 2000

ct_bone = np.clip(ct_float, bone_min, bone_max)

# Normalize to 0-1
ct_normalized = (ct_bone - bone_min) / (bone_max - bone_min)

print("Original CT range:")
print(ct_float.min(), "to", ct_float.max())

print("\nAfter bone window:")
print(ct_bone.min(), "to", ct_bone.max())

print("\nNormalized range:")
print(ct_normalized.min(), "to", ct_normalized.max())
```

```python
print("CT shape:", ct.shape)

print("Original CT range:")
print("Min:", ct.min())
print("Max:", ct.max())

print("\nNormalized CT range:")
print("Min:", ct_normalized.min())
print("Max:", ct_normalized.max())

# Find slices containing the most bone-window pixels
bone_pixels_per_slice = np.sum(ct_normalized > 0.20, axis=(1, 2))

best_slice = np.argmax(bone_pixels_per_slice)

print("\nSlice with most bone pixels:", best_slice)
print("Bone pixels in that slice:", bone_pixels_per_slice[best_slice])
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ct_normalized has shape:
# (slices, height, width)

print("Original CT shape:", ct_normalized.shape)

# Downsample the CT by a factor of 4 in height and width
coarse_ct = ct_normalized[:, ::4, ::4]

print("Coarse CT shape:", coarse_ct.shape)
```

```python
coarse_slice = coarse_ct[slice_index]

plt.figure(figsize=(8, 6))
plt.imshow(coarse_slice, cmap="gray")
plt.title("Coarse CT Representation")
plt.axis("off")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Threshold for bone in the normalized CT
bone_threshold = 0.20

# Create coarse bone mask
coarse_bone_mask = coarse_ct > bone_threshold

print("Coarse mask shape:", coarse_bone_mask.shape)
print("Bone pixels:", np.sum(coarse_bone_mask))

# Display coarse bone mask
plt.figure(figsize=(8, 6))
plt.imshow(coarse_bone_mask[slice_index], cmap="gray")
plt.title("Coarse Bone Mask")
plt.axis("off")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import ndimage

# Remove very small isolated regions
clean_mask = ndimage.binary_opening(
    coarse_bone_mask,
    structure=np.ones((3, 3, 3))
)

# Find all locations containing coarse bone
coords = np.argwhere(clean_mask)

z_min, y_min, x_min = coords.min(axis=0)
z_max, y_max, x_max = coords.max(axis=0)

print("Initial ROI:")
print("Z:", z_min, "to", z_max)
print("Y:", y_min, "to", y_max)
print("X:", x_min, "to", x_max)
```

```python
# Add margin around ROI
margin_z = 10
margin_y = 10
margin_x = 10

z_min = max(0, z_min - margin_z)
y_min = max(0, y_min - margin_y)
x_min = max(0, x_min - margin_x)

z_max = min(coarse_ct.shape[0], z_max + margin_z)
y_max = min(coarse_ct.shape[1], y_max + margin_y)
x_max = min(coarse_ct.shape[2], x_max + margin_x)

print("\nFinal coarse ROI:")
print("Z:", z_min, "to", z_max)
print("Y:", y_min, "to", y_max)
print("X:", x_min, "to", x_max)
```

```python
coarse_roi = coarse_ct[
    z_min:z_max,
    y_min:y_max,
    x_min:x_max
]

print("\nOriginal coarse volume:", coarse_ct.shape)
print("Coarse ROI:", coarse_roi.shape)
```

```python
roi_slice = coarse_roi.shape[0] // 2

plt.figure(figsize=(8, 6))
plt.imshow(coarse_roi[roi_slice], cmap="gray")
plt.title("Automatically Detected Bone ROI")
plt.axis("off")
plt.show()
```

```python
# Downsampling factor used earlier
scale_y = 4
scale_x = 4

# Convert coarse ROI coordinates back to original CT coordinates
orig_y_min = y_min * scale_y
orig_y_max = min(y_max * scale_y, ct_normalized.shape[1])

orig_x_min = x_min * scale_x
orig_x_max = min(x_max * scale_x, ct_normalized.shape[2])

# Z was not downsampled
orig_z_min = z_min
orig_z_max = min(z_max, ct_normalized.shape[0])

print("Original-resolution ROI:")
print("Z:", orig_z_min, "to", orig_z_max)
print("Y:", orig_y_min, "to", orig_y_max)
print("X:", orig_x_min, "to", orig_x_max)
```

```python
high_res_roi = ct_normalized[
    orig_z_min:orig_z_max,
    orig_y_min:orig_y_max,
    orig_x_min:orig_x_max
]

print("Original CT:", ct_normalized.shape)
print("High-resolution ROI:", high_res_roi.shape)
```

```python
fine_slice = high_res_roi.shape[0] // 2

plt.figure(figsize=(8, 8))
plt.imshow(high_res_roi[fine_slice], cmap="gray")
plt.title("High-Resolution ROI")
plt.axis("off")
plt.show()
```

```python
# Add a larger safety margin in the Z direction
safe_margin_z = 45

safe_z_min = max(0, z_min - safe_margin_z)
safe_z_max = min(coarse_ct.shape[0], z_max + safe_margin_z)

# Convert Y and X back to original resolution
orig_y_min = y_min * 4
orig_y_max = min(y_max * 4, ct_normalized.shape[1])

orig_x_min = x_min * 4
orig_x_max = min(x_max * 4, ct_normalized.shape[2])

# Extract safer high-resolution ROI
high_res_roi = ct_normalized[
    safe_z_min:safe_z_max,
    orig_y_min:orig_y_max,
    orig_x_min:orig_x_max
]

print("Safe high-resolution ROI:", high_res_roi.shape)
print("Z:", safe_z_min, "to", safe_z_max)
print("Y:", orig_y_min, "to", orig_y_max)
print("X:", orig_x_min, "to", orig_x_max)
```

```python
fine_slice = high_res_roi.shape[0] // 2

plt.figure(figsize=(8, 8))
plt.imshow(high_res_roi[fine_slice], cmap="gray")
plt.title("Safe High-Resolution ROI")
plt.axis("off")
plt.show()
```

```python
# Extract the ground-truth label corresponding to our safe ROI

label_roi = label[
    safe_z_min:safe_z_max,
    orig_y_min:orig_y_max,
    orig_x_min:orig_x_max
]

# Count all ground-truth labeled voxels
total_label_voxels = np.sum(label > 0)

# Count labeled voxels captured inside our ROI
roi_label_voxels = np.sum(label_roi > 0)

coverage = (roi_label_voxels / total_label_voxels) * 100

print("Total ground-truth labeled voxels:", total_label_voxels)
print("Labeled voxels inside ROI:", roi_label_voxels)
print("ROI coverage:", round(coverage, 2), "%")
```

```python
# Larger safety margins at original CT resolution
extra_z = 20
extra_y = 40
extra_x = 40

# Expand the current original-resolution ROI
safe_z_min2 = max(0, safe_z_min - extra_z)
safe_z_max2 = min(ct_normalized.shape[0], safe_z_max + extra_z)

safe_y_min2 = max(0, orig_y_min - extra_y)
safe_y_max2 = min(ct_normalized.shape[1], orig_y_max + extra_y)

safe_x_min2 = max(0, orig_x_min - extra_x)
safe_x_max2 = min(ct_normalized.shape[2], orig_x_max + extra_x)

print("Expanded ROI:")
print("Z:", safe_z_min2, "to", safe_z_max2)
print("Y:", safe_y_min2, "to", safe_y_max2)
print("X:", safe_x_min2, "to", safe_x_max2)
```

```python
high_res_roi2 = ct_normalized[
    safe_z_min2:safe_z_max2,
    safe_y_min2:safe_y_max2,
    safe_x_min2:safe_x_max2
]

label_roi2 = label[
    safe_z_min2:safe_z_max2,
    safe_y_min2:safe_y_max2,
    safe_x_min2:safe_x_max2
]

print("New high-resolution ROI:", high_res_roi2.shape)
```

```python
total_label_voxels = np.sum(label > 0)
roi_label_voxels2 = np.sum(label_roi2 > 0)

coverage2 = 100 * roi_label_voxels2 / total_label_voxels

print("Total labeled voxels:", total_label_voxels)
print("Labeled voxels inside new ROI:", roi_label_voxels2)
print("New ROI coverage:", round(coverage2, 2), "%")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Middle slice of the new ROI
roi_slice = high_res_roi2.shape[0] // 2

plt.figure(figsize=(9, 8))

plt.imshow(
    high_res_roi2[roi_slice],
    cmap="gray"
)

masked_label = np.ma.masked_where(
    label_roi2[roi_slice] == 0,
    label_roi2[roi_slice]
)

plt.imshow(
    masked_label,
    cmap="jet",
    alpha=0.45
)

plt.title("Final High-Resolution ROI + Ground Truth")
plt.axis("off")
plt.show()
```

```python
import glob
import os
import random

# Get all CT files
ct_files = (
    glob.glob("/content/PENGWIN/images_part1/*.mha") +
    glob.glob("/content/PENGWIN/images_part2/*.mha")
)

# Sort for reproducibility
ct_files = sorted(ct_files)

# Create case IDs
case_ids = [os.path.basename(f).replace(".mha", "") for f in ct_files]

# Fixed random seed
random.seed(42)
random.shuffle(case_ids)

# Split
train_ids = case_ids[:70]
val_ids   = case_ids[70:85]
test_ids  = case_ids[85:100]

print("Training cases:", len(train_ids))
print("Validation cases:", len(val_ids))
print("Test cases:", len(test_ids))

print("\nTraining:", train_ids[:10])
print("Validation:", val_ids)
print("Testing:", test_ids)
```

```python
import os
import json

# Base folders
part1_dir = "/content/PENGWIN/images_part1"
part2_dir = "/content/PENGWIN/images_part2"
label_dir = "/content/PENGWIN/labels"

# Function to find CT file
def find_ct(case_id):
    path1 = os.path.join(part1_dir, case_id + ".mha")
    path2 = os.path.join(part2_dir, case_id + ".mha")

    if os.path.exists(path1):
        return path1
    elif os.path.exists(path2):
        return path2
    else:
        return None

# Create dataset index
dataset_index = []

for case_id in sorted(train_ids + val_ids + test_ids):

    ct_path = find_ct(case_id)
    label_path = os.path.join(label_dir, case_id + ".mha")

    dataset_index.append({
        "case_id": case_id,
        "ct": ct_path,
        "label": label_path
    })

# Save index
with open("/content/PENGWIN/dataset_index.json", "w") as f:
    json.dump(dataset_index, f, indent=4)

print("Dataset index created!")
print("Total cases:", len(dataset_index))

# Display first 5
for item in dataset_index[:5]:
    print(item)
```

```python
import SimpleITK as sitk
import numpy as np

case_id = train_ids[0]

# Find CT path
ct_path = find_ct(case_id)
label_path = os.path.join(label_dir, case_id + ".mha")

# Read CT and label
ct_img = sitk.ReadImage(ct_path)
label_img = sitk.ReadImage(label_path)

# Convert to NumPy
ct = sitk.GetArrayFromImage(ct_img)
label = sitk.GetArrayFromImage(label_img)

print("Case ID:", case_id)

print("\n--- CT ---")
print("Shape:", ct.shape)
print("Spacing:", ct_img.GetSpacing())
print("Data type:", ct.dtype)
print("Minimum:", ct.min())
print("Maximum:", ct.max())

print("\n--- LABEL ---")
print("Shape:", label.shape)
print("Data type:", label.dtype)
print("Unique labels:", np.unique(label))

print("\nSame shape:", ct.shape == label.shape)
```

```python
import numpy as np
from scipy import ndimage

# --------------------------------
# 1. Bone window preprocessing
# --------------------------------

bone_min = 150
bone_max = 2000

ct_float = ct.astype(np.float32)

ct_bone = np.clip(ct_float, bone_min, bone_max)

ct_normalized = (
    (ct_bone - bone_min)
    / (bone_max - bone_min)
)

print("Original CT shape:", ct_normalized.shape)
print("Normalized range:",
      ct_normalized.min(),
      "to",
      ct_normalized.max())


# --------------------------------
# 2. Coarse downsampling
# --------------------------------

downsample_factor = 4

coarse_ct = ct_normalized[
    :,
    ::downsample_factor,
    ::downsample_factor
]

print("Coarse CT shape:", coarse_ct.shape)


# --------------------------------
# 3. Coarse bone detection
# --------------------------------

bone_threshold = 0.20

coarse_bone_mask = coarse_ct > bone_threshold

print("Coarse bone voxels:",
      np.sum(coarse_bone_mask))


# --------------------------------
# 4. Remove small isolated regions
# --------------------------------

coarse_bone_mask = ndimage.binary_opening(
    coarse_bone_mask,
    structure=np.ones((3, 3, 3))
)

print("Bone voxels after cleanup:",
      np.sum(coarse_bone_mask))
```

```python
print("Current CT shape:", ct.shape)
print("Current CT dtype:", ct.dtype)
print("Current CT min:", ct.min())
print("Current CT max:", ct.max())
```

```python
import numpy as np

# Make sure the coarse mask exists
print("Coarse mask shape:", coarse_bone_mask.shape)

# Find all bone voxel coordinates
z, y, x = np.where(coarse_bone_mask)

if len(z) == 0:
    raise ValueError("No bone region detected. We need to adjust the threshold.")

# Bounding box in COARSE coordinates
z_min, z_max = z.min(), z.max()
y_min, y_max = y.min(), y.max()
x_min, x_max = x.min(), x.max()

print("\nCoarse ROI before margin:")
print("Z:", z_min, "to", z_max)
print("Y:", y_min, "to", y_max)
print("X:", x_min, "to", x_max)

# Add safety margin in coarse space
margin_z = 5
margin_y = 8
margin_x = 8

z_min = max(0, z_min - margin_z)
z_max = min(coarse_bone_mask.shape[0] - 1, z_max + margin_z)

y_min = max(0, y_min - margin_y)
y_max = min(coarse_bone_mask.shape[1] - 1, y_max + margin_y)

x_min = max(0, x_min - margin_x)
x_max = min(coarse_bone_mask.shape[2] - 1, x_max + margin_x)

print("\nCoarse ROI after margin:")
print("Z:", z_min, "to", z_max)
print("Y:", y_min, "to", y_max)
print("X:", x_min, "to", x_max)
```

```python
downsample_factor = 4
```

```python
# -----------------------------------------
# Map coarse ROI → original CT coordinates
# -----------------------------------------

factor = downsample_factor

# Z was NOT downsampled
high_z_min = z_min
high_z_max = z_max

# Y and X were downsampled by factor 4
high_y_min = y_min * factor
high_y_max = min((y_max + 1) * factor, ct.shape[1])

high_x_min = x_min * factor
high_x_max = min((x_max + 1) * factor, ct.shape[2])

print("High-resolution ROI:")
print("Z:", high_z_min, "to", high_z_max)
print("Y:", high_y_min, "to", high_y_max)
print("X:", high_x_min, "to", high_x_max)

# Extract high-resolution ROI
high_res_roi = ct[
    high_z_min:high_z_max + 1,
    high_y_min:high_y_max,
    high_x_min:high_x_max
]

print("\nOriginal CT shape:", ct.shape)
print("High-resolution ROI shape:", high_res_roi.shape)

# Calculate voxel reduction
original_voxels = np.prod(ct.shape)
roi_voxels = np.prod(high_res_roi.shape)

reduction = (1 - roi_voxels / original_voxels) * 100

print("Original voxels:", original_voxels)
print("ROI voxels:", roi_voxels)
print("Voxel reduction:", round(reduction, 2), "%")
```

```python
import numpy as np
from scipy import ndimage

# ---------------------------------------
# Coarse bone mask
# ---------------------------------------

threshold = 0.20

coarse_mask = coarse_ct > threshold

print("Coarse mask:", coarse_mask.shape)
print("Bone voxels:", np.sum(coarse_mask))


# ---------------------------------------
# 3D connected components
# ---------------------------------------

structure = ndimage.generate_binary_structure(
    3,      # 3D
    2       # connectivity
)

components, num_components = ndimage.label(
    coarse_mask,
    structure=structure
)

print("Number of connected components:",
      num_components)


# ---------------------------------------
# Component sizes
# ---------------------------------------

component_sizes = np.bincount(
    components.ravel()
)

# Ignore background
component_sizes[0] = 0

# Sort components by size
largest_indices = np.argsort(
    component_sizes
)[::-1]

print("\nLargest components:")

for i in range(min(15, len(largest_indices))):

    component_id = largest_indices[i]

    print(
        f"{i+1}. "
        f"Component {component_id}: "
        f"{component_sizes[component_id]} voxels"
    )
```

```python
# ---------------------------------------
# Select the 4 largest components
# ---------------------------------------

num_keep = 4

selected_components = largest_indices[:num_keep]

print("Selected components:",
      selected_components)

# Create combined mask
selected_mask = np.isin(
    components,
    selected_components
)

print("Selected bone voxels:",
      np.sum(selected_mask))


# ---------------------------------------
# Find bounding box
# ---------------------------------------

z, y, x = np.where(selected_mask)

z_min4, z_max4 = z.min(), z.max()
y_min4, y_max4 = y.min(), y.max()
x_min4, x_max4 = x.min(), x.max()

print("\nROI before margin:")
print("Z:", z_min4, "to", z_max4)
print("Y:", y_min4, "to", y_max4)
print("X:", x_min4, "to", x_max4)


# ---------------------------------------
# Add safety margins
# ---------------------------------------

margin_z = 5
margin_y = 8
margin_x = 8

z_min4 = max(0, z_min4 - margin_z)
z_max4 = min(
    coarse_mask.shape[0] - 1,
    z_max4 + margin_z
)

y_min4 = max(0, y_min4 - margin_y)
y_max4 = min(
    coarse_mask.shape[1] - 1,
    y_max4 + margin_y
)

x_min4 = max(0, x_min4 - margin_x)
x_max4 = min(
    coarse_mask.shape[2] - 1,
    x_max4 + margin_x
)

print("\nROI after margin:")
print("Z:", z_min4, "to", z_max4)
print("Y:", y_min4, "to", y_max4)
print("X:", x_min4, "to", x_max4)
```
