```python
import SimpleITK as sitk
import numpy as np

from monai.transforms import (
    MapTransform,
    Compose,
    EnsureChannelFirstd,
    Orientationd,
    Spacingd,
    ScaleIntensityRanged,
    RandCropByPosNegLabeld,
)


class LoadMHAd(MapTransform):
    """
    Load .mha files using SimpleITK.
    """

    def __call__(self, data):
        d = dict(data)

        for key in self.keys:
            image = sitk.ReadImage(d[key])
            array = sitk.GetArrayFromImage(image)

            d[key] = array

            # Store spacing for later transforms
            d[f"{key}_spacing"] = image.GetSpacing()

        return d


print("Custom MHA loader created successfully.")
```

```python
import SimpleITK as sitk
import numpy as np
from monai.transforms import MapTransform


class LoadMHAd(MapTransform):
    """
    Load PENGWIN .mha files using SimpleITK
    and explicitly add the channel dimension.
    """

    def __call__(self, data):
        d = dict(data)

        for key in self.keys:
            image = sitk.ReadImage(d[key])
            array = sitk.GetArrayFromImage(image)

            # SimpleITK gives: Z, Y, X
            # Add channel dimension: 1, Z, Y, X
            array = np.expand_dims(array, axis=0)

            d[key] = array

        return d


print("Fixed SimpleITK MHA loader created successfully.")
```

```python
import SimpleITK as sitk
import numpy as np
from monai.data import MetaTensor
from monai.transforms import MapTransform


class LoadMHAd(MapTransform):
    """
    Load PENGWIN .mha files using SimpleITK
    while preserving physical metadata.
    """

    def __call__(self, data):
        d = dict(data)

        for key in self.keys:

            image = sitk.ReadImage(d[key])

            array = sitk.GetArrayFromImage(image).astype(np.float32)

            # SimpleITK array order:
            # Z, Y, X
            #
            # Add channel:
            # C, Z, Y, X
            array = np.expand_dims(array, axis=0)

            # SimpleITK spacing is:
            # X, Y, Z
            spacing = image.GetSpacing()

            # Store original metadata
            d[key] = MetaTensor(
                array,
                meta={
                    "filename_or_obj": d[key],
                    "spacing": spacing,
                    "original_spacing": spacing,
                    "spatial_shape": array.shape[1:]
                }
            )

        return d


print("Metadata-preserving MHA loader created.")
```

```python
import time
import torch

# --------------------------------------------------
# Short training test
# --------------------------------------------------

TEST_EPOCHS = 2

for epoch in range(TEST_EPOCHS):

    model.train()

    epoch_loss = 0.0
    start_time = time.time()

    for step, batch in enumerate(train_loader):

        # Move data to GPU
        images = batch["image"].to(device)
        labels = batch["label"].long().to(device)

        # Clear gradients
        optimizer.zero_grad(set_to_none=True)

        # Mixed precision forward pass
        with torch.amp.autocast("cuda"):

            outputs = model(images)

            loss = loss_function(
                outputs,
                labels
            )

        # Backpropagation
        scaler.scale(loss).backward()

        # Optimizer update
        scaler.step(optimizer)
        scaler.update()

        epoch_loss += loss.item()

        # Print progress
        if (step + 1) % 10 == 0:
            print(
                f"Epoch {epoch + 1}/{TEST_EPOCHS} "
                f"| Step {step + 1}/{len(train_loader)} "
                f"| Loss: {loss.item():.4f}"
            )

    epoch_loss /= len(train_loader)

    elapsed = time.time() - start_time

    print(
        f"\nEpoch {epoch + 1} completed "
        f"| Average Loss: {epoch_loss:.4f} "
        f"| Time: {elapsed/60:.2f} min\n"
    )

print("2-epoch training test completed successfully.")
```

```python
import numpy as np
import SimpleITK as sitk
import torch

from monai.data import MetaTensor
from monai.transforms import MapTransform


class LoadMHAd(MapTransform):
    """
    Load PENGWIN .mha files using SimpleITK
    while preserving basic physical metadata.
    """

    def __call__(self, data):
        d = dict(data)

        for key in self.keys:
            image = sitk.ReadImage(d[key])

            # SimpleITK gives Z,Y,X
            array = sitk.GetArrayFromImage(image).astype(np.float32)

            # Add channel dimension -> C,Z,Y,X
            array = np.expand_dims(array, axis=0)

            spacing = image.GetSpacing()

            d[key] = MetaTensor(
                array,
                meta={
                    "filename_or_obj": d[key],
                    "spacing": spacing,
                    "original_spacing": spacing,
                    "spatial_shape": array.shape[1:]
                }
            )

        return d


print("LoadMHAd created successfully.")
```

```python
from monai.transforms import (
    Compose,
    Orientationd,
    Spacingd,
    ScaleIntensityRanged,
    RandCropByPosNegLabeld
)

train_transforms = Compose([
    LoadMHAd(keys=["image", "label"]),

    # Standardize orientation
    Orientationd(
        keys=["image", "label"],
        axcodes="RAS"
    ),

    # Resample to 1 mm × 1 mm × 1 mm
    Spacingd(
        keys=["image", "label"],
        pixdim=(1.0, 1.0, 1.0),
        mode=("bilinear", "nearest")
    ),

    # CT intensity normalization
    ScaleIntensityRanged(
        keys=["image"],
        a_min=-200,
        a_max=2000,
        b_min=0.0,
        b_max=1.0,
        clip=True
    ),

    # 96³ training patches
    RandCropByPosNegLabeld(
        keys=["image", "label"],
        label_key="label",
        spatial_size=(96, 96, 96),
        pos=1,
        neg=1,
        num_samples=2,
        image_key="image",
        image_threshold=0
    )
])


val_transforms = Compose([
    LoadMHAd(keys=["image", "label"]),

    Orientationd(
        keys=["image", "label"],
        axcodes="RAS"
    ),

    Spacingd(
        keys=["image", "label"],
        pixdim=(1.0, 1.0, 1.0),
        mode=("bilinear", "nearest")
    ),

    ScaleIntensityRanged(
        keys=["image"],
        a_min=-200,
        a_max=2000,
        b_min=0.0,
        b_max=1.0,
        clip=True
    )
])

print("Training transforms created.")
print("Validation transforms created.")
```

```python
from monai.data import Dataset

BASE = "/content/PENGWIN"
CT_DIR = f"{BASE}/CT"
LABEL_DIR = f"{BASE}/Labels"

train_files = [
    {
        "image": f"{CT_DIR}/{case_id}.mha",
        "label": f"{LABEL_DIR}/{case_id}.mha"
    }
    for case_id in train_ids
]

val_files = [
    {
        "image": f"{CT_DIR}/{case_id}.mha",
        "label": f"{LABEL_DIR}/{case_id}.mha"
    }
    for case_id in val_ids
]

test_files = [
    {
        "image": f"{CT_DIR}/{case_id}.mha",
        "label": f"{LABEL_DIR}/{case_id}.mha"
    }
    for case_id in test_ids
]

print("Train files:", len(train_files))
print("Validation files:", len(val_files))
print("Test files:", len(test_files))

print("\nExample:")
print(train_files[0])
```

```python
from monai.data import Dataset

train_ds = Dataset(
    data=train_files,
    transform=train_transforms
)

val_ds = Dataset(
    data=val_files,
    transform=val_transforms
)

print("Train dataset:", len(train_ds))
print("Validation dataset:", len(val_ds))
```

```python
sample = train_ds[0]

print("Number of patches returned:", len(sample))

print("\nImage shape:")
print(sample[0]["image"].shape)

print("\nLabel shape:")
print(sample[0]["label"].shape)

print("\nImage dtype:")
print(sample[0]["image"].dtype)

print("\nLabel dtype:")
print(sample[0]["label"].dtype)

print("\nImage range:")
print(
    float(sample[0]["image"].min()),
    "to",
    float(sample[0]["image"].max())
)

print("\nLabels present:")
print(torch.unique(sample[0]["label"]).tolist())
```

```python
from torch.utils.data import DataLoader

train_loader = DataLoader(
    train_ds,
    batch_size=1,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=1,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

print("Train batches:", len(train_loader))
print("Validation batches:", len(val_loader))
```

```python
batch = next(iter(train_loader))

print("Batch type:", type(batch))
print("Number of patches:", len(batch))

print("\nFirst patch image shape:")
print(batch[0]["image"].shape)

print("First patch label shape:")
print(batch[0]["label"].shape)

print("\nSecond patch image shape:")
print(batch[1]["image"].shape)

print("Second patch label shape:")
print(batch[1]["label"].shape)

print("\nLabels in first patch:")
print(torch.unique(batch[0]["label"]).tolist())

print("Labels in second patch:")
print(torch.unique(batch[1]["label"]).tolist())
```

```python
import torch
from monai.networks.nets import UNet

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = UNet(
    spatial_dims=3,
    in_channels=1,
    out_channels=31,          # background + labels 1–30
    channels=(16, 32, 64, 128),
    strides=(2, 2, 2),
    num_res_units=1,
    norm="INSTANCE"
).to(device)

print("Device:", device)
print("Model created successfully.")

total_params = sum(p.numel() for p in model.parameters())
print("Total parameters:", total_params)
```

```python
model.eval()

with torch.inference_mode():
    test_input = batch[0]["image"].to(device)
    test_output = model(test_input)

print("Input shape :", test_input.shape)
print("Output shape:", test_output.shape)
print("Output dtype:", test_output.dtype)

print(
    "GPU memory allocated:",
    round(torch.cuda.memory_allocated() / 1024**3, 3),
    "GB"
)
```

```python
from monai.losses import DiceCELoss

loss_function = DiceCELoss(
    include_background=False,
    to_onehot_y=True,
    softmax=True,
    lambda_dice=1.0,
    lambda_ce=1.0
)

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=2e-4,
    weight_decay=1e-5
)

print("Loss function:", loss_function)
print("Optimizer:", optimizer.__class__.__name__)
print("Learning rate:", 2e-4)
```

```python
model.train()

# Get one batch
batch = next(iter(train_loader))

# Because num_samples=2, DataLoader returns a list of patches
patch = batch[0]

images = patch["image"].to(device)
labels = patch["label"].to(device)

# Remove unnecessary channel dimension from labels
# [B, 1, Z, Y, X] → [B, Z, Y, X]
labels = labels.long()

# Forward pass
optimizer.zero_grad(set_to_none=True)

outputs = model(images)

# Calculate loss
loss = loss_function(outputs, labels)

# Backpropagation
loss.backward()

# Update model
optimizer.step()

print("Training step completed successfully.")
print("Input shape :", images.shape)
print("Output shape:", outputs.shape)
print("Label shape :", labels.shape)
print("Loss        :", round(loss.item(), 4))

print(
    "GPU memory allocated:",
    round(torch.cuda.memory_allocated() / 1024**3, 3),
    "GB"
)
```

```python
import torch
from monai.networks.nets import UNet

# Clear the test model
del model
del optimizer

torch.cuda.empty_cache()

# Create a fresh model
model = UNet(
    spatial_dims=3,
    in_channels=1,
    out_channels=31,
    channels=(16, 32, 64, 128),
    strides=(2, 2, 2),
    num_res_units=1,
    norm="INSTANCE"
).to(device)

# Fresh optimizer
optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=2e-4,
    weight_decay=1e-5
)

print("Fresh baseline model created.")

print(
    "Parameters:",
    sum(p.numel() for p in model.parameters())
)

print(
    "GPU memory allocated:",
    round(torch.cuda.memory_allocated() / 1024**3, 3),
    "GB"
)
```

```python
import time
import torch

num_epochs = 5

for epoch in range(num_epochs):

    model.train()
    epoch_loss = 0.0
    start_time = time.time()

    for step, batch in enumerate(train_loader, start=1):

        # DataLoader returns 2 patches because num_samples=2
        for patch in batch:

            images = patch["image"].to(device, non_blocking=True)
            labels = patch["label"].to(device, non_blocking=True).long()

            optimizer.zero_grad(set_to_none=True)

            outputs = model(images)

            loss = loss_function(outputs, labels)

            loss.backward()
            optimizer.step()

            epoch_loss += loss.item()

            # Release patch tensors before next iteration
            del images, labels, outputs, loss

    avg_loss = epoch_loss / (len(train_loader) * 2)
    elapsed = time.time() - start_time

    print(
        f"Epoch [{epoch+1}/{num_epochs}] "
        f"Loss: {avg_loss:.4f} "
        f"Time: {elapsed/60:.2f} min"
    )

    torch.cuda.empty_cache()
```

```python
import numpy as np
import torch
import gc

from monai.inferers import sliding_window_inference

# Use the first validation case
val_case = val_files[0]

print("Validation case:", val_case["image"])

# Load and preprocess ONE case
case = val_transforms(
    {
        "image": val_case["image"],
        "label": val_case["label"]
    }
)

image = case["image"]

print("Preprocessed image shape:", image.shape)

# Add batch dimension
image = image.unsqueeze(0)

print("Inference input shape:", image.shape)

model.eval()

with torch.inference_mode():

    prediction = sliding_window_inference(
        inputs=image,
        roi_size=(96, 96, 96),
        sw_batch_size=1,
        predictor=model,
        overlap=0.0,
        mode="constant",
        sw_device="cuda",
        device="cpu"
    )

print("Prediction created on CPU.")
print("Prediction shape:", prediction.shape)
print("Prediction dtype:", prediction.dtype)

# Immediately convert to class labels on CPU
pred_labels = torch.argmax(prediction, dim=1)

print("Class prediction shape:", pred_labels.shape)
print(
    "Predicted labels:",
    torch.unique(pred_labels).tolist()
)

# Release the large prediction immediately
del prediction
gc.collect()
torch.cuda.empty_cache()

print(
    "GPU memory allocated:",
    round(torch.cuda.memory_allocated() / 1024**3, 3),
    "GB"
)
```
